```python
# Import relevant packages
import numpy as np
import pandas as pd
import datetime
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.cm as cm
import seaborn as sns
import os
import math
```

### Read in processed files 

```python
# Read in processed data files
df_energy_clean = pd.read_csv('../data/processed/energy_dataset_clean.csv')
df_weather_wide = pd.read_csv('../data/processed/weather_wide.csv')
```

### Grouping generation sources by type 

#### feel free to change the groupings 

```python
solar_cols        = ['generation solar']
wind_cols         = ['generation wind onshore']
hydro_cols        = ['generation hydro run-of-river and poundage',
                     'generation hydro water reservoir']
pumped_hydro_cols = ['generation hydro pumped storage consumption']
fossil_cols       = ['generation fossil brown coal/lignite',
                     'generation fossil gas',
                     'generation fossil hard coal',
                     'generation fossil oil']
nuclear_cols      = ['generation nuclear']
other_cols        = ['generation other', 'generation other renewable', 
                     'generation biomass', 'generation waste']  # biomass added, generation waste added

df = df_energy_clean.copy()

df['gen_solar']        = df[solar_cols].sum(axis=1)
df['gen_wind']         = df[wind_cols].sum(axis=1)
df['gen_hydro']        = df[hydro_cols].sum(axis=1)
df['gen_pumped_hydro'] = df[pumped_hydro_cols].sum(axis=1)
df['gen_fossil']       = df[fossil_cols].sum(axis=1)
df['gen_nuclear']      = df[nuclear_cols].sum(axis=1)
df['gen_other']        = df[other_cols].sum(axis=1)

gen_cols_to_drop = (solar_cols + wind_cols + hydro_cols + pumped_hydro_cols +
                    fossil_cols + nuclear_cols + other_cols)
df.drop(columns=gen_cols_to_drop, inplace=True)

df_energy_clean = df.copy()
```

```python
df_energy_clean
```

### Merge weather & energy dfs 

```python
df_energy_weather = df_energy_clean.merge(
    df_weather_wide.drop(columns=["date", "hour"]),
    on="time",
    how="inner",
    validate="one_to_one"
)
```

```python
df_energy_weather
```

```python
gen_cols = ['gen_solar', 'gen_wind', 'gen_hydro', 'gen_pumped_hydro', 'gen_fossil', 'gen_nuclear', 'gen_other']
load_cols = ['total load actual']

total_gen_per_hour = df_energy_weather[gen_cols].sum(axis=1).rolling(window=24).mean()
total_load_per_hour = df_energy_weather[load_cols].rolling(window=24).mean()

# Plot total generation and load
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(total_gen_per_hour, label='Total Generation (Daily Average)', alpha=0.7)
ax.plot(total_load_per_hour, label='Total Load (Daily Average)', alpha=0.5)
ax.set_xlabel('Hour', fontweight='bold')
ax.set_ylabel('Generation/Load (MW)', fontweight='bold')
ax.legend()

plt.tight_layout()
plt.show()
```

### Visualize weather features across cities

```python
# Define the cities
cities = ['Barcelona', 'Bilbao', 'Madrid', 'Seville', 'Valencia']

# Define the weather features
weather_feats = [
    'temp',
    'pressure',
    'humidity',
    'wind_speed',
    'wind_deg',
    'rain_1h',
    'rain_3h',
    'snow_3h',
    'clouds_all'
]

# Convert date column to datetime if needed
df_weather_wide['date'] = pd.to_datetime(df_weather_wide['date'])

# Define temporal span
date = df_weather_wide['date']

for feature in weather_feats:

    # Initialize the subplot
    fig, ax = plt.subplots(figsize=(10,6))

    # Plot features for all cities
    for city in cities:

        # Original time series
        feat = df_weather_wide[f'{feature}_{city}']

        # Daily rolling average of time series
        feat_rolling = feat.rolling(window=24).mean()

        # Plot the feature
        line = ax.plot(
            date,
            feat,
            alpha=0.25,
            linewidth=0.5
        )[0]

        # Plot the rolling average
        ax.plot(
            date,
            feat_rolling,
            color=line.get_color(),  # Plot the same color
            linewidth=1,
            label=city
        )
    
    # Set the plot labels
    ax.set_xlabel('Date')
    ax.set_ylabel(feature)
    ax.set_title(feature)
    ax.legend()

    # Date formatting
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=1))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%Y'))

    # Rotate dates
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right')

    # Customize the figure
    plt.tight_layout()
    plt.show()
```

### Visualize energy & weather feature combinations

```python
# Compute average temperature across all cities
temp_cols = ['temp_Barcelona', 'temp_Bilbao', 'temp_Madrid', 'temp_Seville', 'temp_Valencia']
df_energy_weather['temp_avg'] = (
    df_energy_weather[temp_cols].mean(axis=1)
)
```

```python
# Define plotting function
def plot_dual_axs(df, left_cols, right_cols, window, ax_left_label, ax_right_label, fig_title):

    # Convert date column to datetime if needed
    df['date'] = pd.to_datetime(df['date'])

    # Define temporal span
    date = df['date']

    # Intialize the plot
    fig, ax1 = plt.subplots(figsize=(12,6))

    # Create secondary axis
    ax2 = ax1.twinx()

    # Define the colormaps
    left_cmap = cm.get_cmap('tab10', len(left_cols))
    left_colors = [left_cmap(i) for i in range(len(left_cols))]

    right_cmap = cm.get_cmap('YlGn', len(right_cols))
    right_colors = [right_cmap(i) for i in range(len(right_cols))]

    # ---------Plot features on LEFT axis---------
    for (col, l_color) in zip(left_cols, left_colors):

        # Raw time series
        y1_raw = df[col]

        # Rolling average time series
        y1_rolling = y1_raw.rolling(window=window).mean()

        # Plot raw data
        line = ax1.plot(
            date,
            y1_raw,
            color=l_color,
            alpha=0.25,
            linewidth=0.5
        )[0]

        # Plot rolling data
        ax1.plot(
            date,
            y1_rolling,
            color=line.get_color(),  # Plot the same color as raw data
            linewidth=1,
            label=col
        )

    # ---------Plot features on RIGHT axis---------
    for (col, r_color) in zip(right_cols, right_colors):

        # Raw time series
        y2_raw = df[col]

        # Rolling average time series
        y2_rolling = y2_raw.rolling(window=window).mean()

        # Plot raw data
        line = ax2.plot(
            date,
            y2_raw,
            color=r_color,
            alpha=0.25,
            linewidth=0.5
        )[0]

        # Plot rolling data
        ax2.plot(
            date,
            y2_rolling,
            color=line.get_color(),  # Plot the same color as raw data
            linewidth=1,
            linestyle='--',
            label=col
        )

    # Add labels
    ax1.set_xlabel('Date', fontsize=10)

    ax1.set_ylabel(
        ax_left_label,
        fontsize=10
    )

    ax2.set_ylabel(
        ax_right_label,
        fontsize=10
    )

    ax1.set_title(
        fig_title,
        fontsize=12
    )

    # Date formatting
    ax1.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
    ax1.xaxis.set_major_formatter(mdates.DateFormatter('%m-%Y'))

    # Rotate dates
    plt.setp(ax1.get_xticklabels(), rotation=45, ha='right')

    # Combine legends from both axes
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()

    ax1.legend(
        lines1 + lines2,
        labels1 + labels2,
        loc='upper left',
        bbox_to_anchor=(1.06, 1)
    )

    # Customize the figure
    plt.tight_layout()
    plt.show()

    return
```

#### Temperature, Load, Generation

```python
# Define the columns
gen_cols = ['gen_solar',
            'gen_wind',
            'gen_hydro',
            'gen_pumped_hydro',
            'gen_fossil',
            'gen_nuclear',
            'gen_other']

load_cols = ['total load actual']

weather_cols = ['temp_avg']

# Plot the data
plot_dual_axs(
    df=df_energy_weather,
    left_cols=gen_cols + load_cols,
    right_cols=weather_cols,
    window=24,
    ax_left_label='Generation/Load (MW)',
    ax_right_label='Temperature (°C)',
    fig_title='Generation, Load, Temperature'
)
```
